```python
# Simple GPA Calculator

# Subject 1
grade1 = 3.4
credit1 = 3

# Subject 2
grade2 = 2.7
credit2 = 2

# Subject 3
grade3 = 2.4
credit3 = 4
# Subject 4
grade4 = 1.5
credit4 = 3
# Subject 5
grade5 = 1
credit5 = 3
# Subject 6
grade6 = 4
credit6 = 3
# Calculate total points
total_points = (grade1 * credit1) + (grade2 * credit2) + (grade3 * credit3)+(grade4 * credit4)+(grade5 * credit5)+(grade6 * credit6)

# Calculate total credit hours
total_credits = credit1 + credit2 + credit3 + credit4 + credit5 + credit6

# Calculate GPA
gpa = total_points / total_credits

print("Your GPA is:", gpa)
```
